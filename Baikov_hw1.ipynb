```python
# Лабораторна робота 1.Байков Олександр

import numpy as np

# 1. Одновимірний масив з перших 20 натуральних чисел
vec_20 = np.arange(1, 21)
print("1. Вектор з перших 20 натуральних чисел:\n", vec_20)

# 2. Двовимірний масив 4x4, заповнений одиницями
matrix_ones = np.ones((4, 4))
print("\n2. Матриця 4x4, заповнена одиницями:\n", matrix_ones)

# 3. Масив 5x5 з випадковими дійсними числами в діапазоні від -1 до 1
matrix_random_5x5 = np.random.uniform(-1, 1, (5, 5))
print("\n3. Матриця 5x5 з випадковими дійсними числами від -1 до 1:\n", matrix_random_5x5)

# 4. Масив 3x3 з випадковими дійсними числами в діапазоні від 0 до 1
matrix_random_3x3 = np.random.rand(3, 3)
print("\n4. Матриця 3x3 з випадковими дійсними числами від 0 до 1:\n", matrix_random_3x3)

# 5. Два одновимірні масиви розміром 6 з випадковими цілими числами від 1 до 20
arr1 = np.random.randint(1, 21, 6)
arr2 = np.random.randint(1, 21, 6)
print("\n5. Два одновимірні масиви:\n", "arr1 =", arr1, "\narr2 =", arr2)
print("Сума:", arr1 + arr2)
print("Різниця:", arr1 - arr2)
print("Добуток:", arr1 * arr2)

# 6. Дві матриці 3x2 і 2x4 з випадковими цілими числами від 1 до 5, перемноження
A = np.random.randint(1, 6, (3, 2))
B = np.random.randint(1, 6, (2, 4))
C = A @ B
print("\n6. Матриця A (3x2):\n", A)
print("Матриця B (2x4):\n", B)
print("Добуток A @ B (3x4):\n", C)

# 7. Матриця 3x3 з випадковими дійсними числами від 0 до 1 та транспонування
M = np.random.rand(3, 3)
print("\n7. Матриця 3x3:\n", M)
print("Транспонована матриця:\n", M.T)

# 8. Дві матриці 3x3 з випадковими цілими числами від 1 до 10, знаходження різниці
A1 = np.random.randint(1, 11, (3, 3))
A2 = np.random.randint(1, 11, (3, 3))
print("\n8. Матриця A1:\n", A1)
print("Матриця A2:\n", A2)
print("Різниця A1 - A2:\n", A1 - A2)

# 9. Дві матриці 3x3 з випадковими цілими числами від -5 до 5, знаходження добутку
B1 = np.random.randint(-5, 6, (3, 3))
B2 = np.random.randint(-5, 6, (3, 3))
print("\n9. Матриця B1:\n", B1)
print("Матриця B2:\n", B2)
print("Добуток B1 @ B2:\n", B1 @ B2)

# 10. Матриця 3x3 з випадковими дійсними числами від -1 до 1 та вектор-стовпчик сум рядків
M_sum = np.random.uniform(-1, 1, (3, 3))
row_sums = np.sum(M_sum, axis=1).reshape(-1, 1)
print("\n10. Матриця M:\n", M_sum)
print("Вектор-стовпчик сум рядків:\n", row_sums)
```
