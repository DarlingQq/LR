```python
# Завдання 1
df = pd.read_csv("Global YouTube Statistics.csv", encoding='latin1')
```

```python
# Завдання 2
print("Перші 5 рядків:\n", df.head())
```

```python
# Завдання 2
print("\nРозмір датасету:", df.shape)
```

```python
# Завдання 3
print("\nПропущені значення в кожному стовпці:\n", df.isna().sum())
```

```python
# Завдання 4
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].apply(lambda x: x.fillna(x.mean()))
```

```python
cat_cols = ['category', 'Country', 'Abbreviation', 'channel_type', 'created_month']
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Завдання 6
print("\nПропуски після заповнення:\n", df.isna().sum())
```

```python
# Завдання 7
print("\nУнікальних країн:", df['Country'].nunique())
```

```python
# Завдання 8
df['video views'].hist(bins=30, color='yellow', edgecolor='black')
plt.title('Розподіл переглядів')
plt.xlabel('Перегляди')
plt.ylabel('Кількість')
plt.grid(True)
plt.show()
```

```python
# Завдання 9
print("\nМаксимум переглядів:", df['video views'].max())
print("Мінімум переглядів:", df['video views'].min())
print("Середнє переглядів:", df['video views'].mean())

```

```python
# Завдання 10
country_uploads = df.groupby('Country')['uploads'].sum()
print("\nКраїна з найбільшою кількістю завантажених відео:", country_uploads.idxmax())
```

```python
# Завдання 11
max_uploads = df.loc[df['uploads'].idxmax()]
min_uploads = df.loc[df['uploads'].idxmin()]
print("\nНайбільше завантажень у:", max_uploads['Youtuber'], "| Кількість:", max_uploads['uploads'])
print("Найменше завантажень у:", min_uploads['Youtuber'], "| Кількість:", min_uploads['uploads'])
```

```python
# Завдання 1
df = pd.read_csv('bestsellers with categories.csv')
```

```python
# Завдання 2
print("Перші 10 рядків:\n", df.head(10))
```

```python
# Завдання 3
print("\nРозмір датасету:", df.shape)
```

```python
# Завдання 4
df.columns = ['name', 'author', 'user_rating', 'reviews', 'price', 'year', 'genre']
print("\nНові назви колонок:", df.columns)
```

```python
# Завдання 5
print("\nПропущені значення:\n", df.isna().sum())
# Пропусків немає.
```

```python
# Завдання 6
print("\nУнікальні жанри:", df['genre'].unique())
```

```python
# Завдання 7
print("\nМаксимальна ціна:", df['price'].max())
print("Мінімальна ціна:", df['price'].min())
print("Середня ціна:", df['price'].mean())
print("Медіанна ціна:", df['price'].median())
```

```python
# Завдання 8
max_rating = df['user_rating'].max()
print("\nНайвищий рейтинг:", max_rating)
```

```python
# Завдання 8.1
print("Кількість книг з рейтингом", max_rating, ":", df[df['user_rating'] == max_rating].shape[0])

```

```python
# Завдання 8.2
most_reviews = df[df['reviews'] == df['reviews'].max()]
print("\nКнига з найбільшою кількістю відгуків:\n", most_reviews[['name', 'reviews']])
```

```python
# Завдання 8.3
top_2010 = df[df['year'] == 2010]
most_expensive_2010 = top_2010[top_2010['price'] == top_2010['price'].max()]
print("\nНайдорожча книга 2010 року:\n", most_expensive_2010[['name', 'price']])
```

```python
# Завдання 8.4
fiction_2012 = df[(df['year'] == 2012) & (df['genre'] == 'Fiction')]
print("\nFiction книг у 2012 році:", fiction_2012.shape[0])
```

```python
# Завдання 8.5
books_49 = df[(df['user_rating'] == 4.9) & (df['year'].isin([2010, 2011]))]
print("\nКількість книг з рейтингом 4.9 у 2010 та 2011:", books_49.shape[0])
```

```python
# Завдання 9
price_stats = df.groupby('genre')['price'].agg(['min', 'max'])
print("\nМін/Макс ціна по жанрах:\n", price_stats)
```
