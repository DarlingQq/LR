```python
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
import pandas as pd
```

```python
# Завдання 1
df_iris = px.data.iris()
```

```python
fig1 = px.scatter(
    df_iris,
    x="sepal_length",
    y="sepal_width",
    color="species",
    symbol="species",
    size="petal_length",
    title="<b>Дискретні кольори </b>"
)
fig1.update_layout(title_x=0.5)
fig1.show()
```

```python
fig2 = px.scatter(
    df_iris,
    x="sepal_length",
    y="sepal_width",
    color="petal_length",
    symbol="species",
    size="petal_length",
    title="<b>Неперервні кольори</b>"
)
fig2.update_layout(title_x=0.5)
fig2.show()
```

```python
fig3 = px.scatter(
    df_iris,
    x="petal_length",
    y="sepal_length",
    color="species",
    symbol="species",
    title="<b>різні маркери</b>"
)
fig3.update_layout(title_x=0.5)
fig3.show()
```

```python
# Завдання 2
df_tips = px.data.tips()
fig4 = px.scatter(
    df_tips,
    x="total_bill",
    y="tip",
    trendline="ols",
    color="sex",
    title="<b>Залежність чайових від суми рахунку</b>",
    labels={"total_bill": "Сума рахунку", "tip": "Чайові", "sex": "Стать"}
)
fig4.update_layout(title_x=0.5)
fig4.show()
```

```python
# Завдання 3
df_gap = px.data.gapminder().query("year == 1952")
fig5 = px.scatter(
    df_gap,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    log_x=True,
    title="<b>Доходи та тривалість життя у світі (1952)</b>",
    labels={
        "gdpPercap": "Дохід на особу",
        "lifeExp": "Тривалість життя",
        "pop": "Населення",
        "continent": "Континент"
    }
)
fig5.update_layout(title_x=0.5)
fig5.show()

```

```python
# Завдання 4
df_flights = px.data.gapminder().query("country == 'Japan'")
fig6 = px.line(
    df_flights,
    x="year",
    y="pop",
    title="<b>Динаміка населення Японії</b>",
    labels={"year": "Рік", "pop": "Населення"}
)
fig6.update_layout(title_x=0.5)
fig6.show()
```

```python
# Завдання 5
fig8 = px.violin(
    df_iris,
    x="species",
    y="petal_length",
    box=True,
    points="all",
    title="<b>Violin-графік: Довжина пелюстки</b>",
    labels={"species": "Вид", "petal_length": "Довжина пелюстки"},
    color_discrete_sequence=["#9b59b6"]
)
fig8.update_layout(title_x=0.5)
fig8.show()
```

```python
# Завдання 6
x = np.linspace(0, 2 * np.pi, 500)

fig9 = go.Figure()
fig9.add_trace(go.Scatter(x=x, y=np.sin(x), mode='lines', name='sin(x)'))
fig9.update_layout(title="<b>Графік функції sin(x)</b>", title_x=0.5,
                   xaxis_title="x", yaxis_title="sin(x)")
fig9.show()
```

```python

fig10 = go.Figure()
fig10.add_trace(go.Scatter(x=x, y=np.cos(x), mode='lines', name='cos(x)', line=dict(color='red')))
fig10.update_layout(title="<b>Графік функції cos(x)</b>", title_x=0.5,
                    xaxis_title="x", yaxis_title="cos(x)")
fig10.show()
```

```python
x_tan = np.linspace(-np.pi / 2 + 0.1, np.pi / 2 - 0.1, 500)
fig11 = go.Figure()
fig11.add_trace(go.Scatter(x=x_tan, y=np.tan(x_tan), mode='lines', name='tan(x)', line=dict(color='black')))
fig11.update_layout(title="<b>Графік функції tan(x)</b>", title_x=0.5,
                    xaxis_title="x", yaxis_title="tan(x)", yaxis_range=[-10, 10])
fig11.show()
```
